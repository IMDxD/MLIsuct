## 1. What is Google Colab?

[Google Colab](https://colab.research.google.com/) ("Colaboratory") is a free cloud Jupyter notebook:

- **Nothing to install** — Python, `numpy`, `pandas`, `matplotlib`, `sklearn`, `torch` run in the browser.
- **Free GPU/TPU** (with limits) — handy for neural nets later in the course.
- **Easy sharing** — like Google Docs: send a link, comment, co-edit.
- **Google Drive integration** — save notebooks and data to your Drive.

When to use Colab in this course: seminars, homework, quick experiments. When *not* to: private/sensitive data, very long trainings (session may reset after ~12 h or on inactivity).

## 2. Interface tour: two cell types

| Cell type | What it holds | How to create |
|---|---|---|
| **Code** | Python code, runs and shows output underneath | `+ Code` button, or `Ctrl/Cmd + M, B` |
| **Text** | Formatted notes (Markdown) | `+ Text` button, or `Ctrl/Cmd + M, M` |

Key buttons to know (top toolbar):
- `▶ Run cell` (or `Shift + Enter`) — execute the current cell and jump to the next one.
- `+ Code` / `+ Text` — insert a new cell.
- `RAM / Disk` meter (top-right) — memory usage; click for details.
- `Runtime` menu — *Change runtime type* (CPU/GPU/TPU), *Restart*, *Factory reset*.
- `Connect` button (top-right) — must be connected (green check) before code runs.

Files live in the 📁 *Files* tab (left sidebar): upload data, browse `/content`, mount Drive (Section 7).

## 3. Running code: order matters!

A notebook keeps **one shared memory** (kernel). Cells can run in any order, but variables must be defined *before* use.

Golden rules:
1. After opening: *Runtime → Run all* (or run top-to-bottom manually).
2. Number in brackets `[3]` = order of execution. If you see `[ * ]`, the cell is still running.
3. If something is weird: *Runtime → Restart and run all*.

### Must-know shortcuts
- `Shift + Enter` — run cell, go to next
- `Ctrl/Cmd + Enter` — run cell, stay
- `Ctrl/Cmd + M, B` — new code cell below · `Ctrl/Cmd + M, M` — turn cell into Markdown
- `Ctrl/Cmd + M, D` — delete cell · `Ctrl/Cmd + M, Z` — undo delete
- `Ctrl + /` (or `Cmd + /`) — comment/uncomment code
- `Tab` — autocomplete · `Shift + Tab` — quick help for a function

## 5. Magic commands: `%`, `%%`, `!`

- `%...` — one-line magic (e.g. `%time`, `%matplotlib inline`).
- `%%...` — whole-cell magic, must be the **first line** (e.g. `%%time`, `%%bash`).
- `!...` — run a shell command (e.g. `!pip install ...`, `!ls`, `!nvidia-smi`).

Most useful: `%pip install <pkg>`, `%matplotlib inline`, `%time <stmt>`, `%%time`, `!ls`, `!pwd`, `%env`, `%load_ext autoreload`.

````python
# Line magic: time one statement
%time sum(range(100_000))

# Shell: where are we, what files are around?
!pwd && ls
````

## 10. Small Colab extras worth knowing

- **Forms:** add `@param` sliders/inputs to play with values without editing code — e.g. `# @param {type:"slider", min:0, max:10, step:1}`.
- **Table display:** dataframes render as scrollable interactive tables automatically.
- **Snippets & templates:** left sidebar → `<>` snippets (e.g. ready-made chart or form code).
- **%matplotlib inline** is default in Colab — plots just appear under the cell.
- **Secrets (🔑 sidebar):** store API keys safely instead of pasting them into code.

````python
# Demo of a Colab form: change n_points in the sidebar form (Colab) or just edit the variable (Jupyter).
# @param {type:"slider", min:10, max:200, step:10}
n_points = 50
import numpy as np
data = np.random.default_rng(0).normal(size=n_points)
print(f"Generated {n_points} points, mean = {data.mean():.3f}")
# ✏️ Task 6: move the slider (Colab) or change n_points, re-run, watch the output change.
````

## 11. Limits and best practices

- Sessions reset after ~12 h or ~90 min of inactivity — **save notebooks to Drive/GitHub**, keep data + code to reproduce results.
- `/content` is scratch space; Drive is persistent storage.
- Pin package versions for reproducibility (`%pip install numpy==1.26.4`).
- Set a random seed (`np.random.seed(...)`, `torch.manual_seed(...)`) so runs repeat.
- Heavy job? Save checkpoints to Drive during training, not only at the end.

### Troubleshooting quick table

| Symptom | Fix |
|---|---|
| `ModuleNotFoundError` | `%pip install <pkg>`, then restart runtime |
| Stale variables / weird errors | *Runtime → Restart and run all* |
| GPU out of memory | Reduce batch size, *Runtime → Factory reset*, use smaller model |
| Drive not mounted | Re-run mount cell, complete the auth link |
| Notebook won't open from GitHub | *File → Upload* the `.ipynb` manually |
